- In a Markdown cell, answer: how is all() different from any()?

```python
list_test = [0, 1, 2, 3, 4, 5, 6, 7, 8]

len(list_test)
```

```python
any(list_test)
```

```python
all(list_test)
```

All will look through the list and only return true if every value in the list has a true value (like a value, a boolean true, etc.) and will return false otherwise. On the other hand, any will look through the list and will return true even if only one value is true, and it will only return false if nothing within the list is true.

# 5. Functions
- Define a function that takes an argument and uses control flow inside it (an if/else or a for loop)
- Call your function and show the result

```python
numbers = [1, 2, 3, 5, 6, 8, 9, 13, 15, 16, 18, 22, 23]

def return_even(num):
    even_numbers = []
    for n in num:
        if n % 2 == 0:
            even_numbers.append(n)
    return even_numbers
    
return_even(numbers)
```
